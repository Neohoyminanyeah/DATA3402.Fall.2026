*Exercise 6:* Modify Exercise 3 to show **row and column labels** so that players can specify locations like `"A2"` or `"C1"`.

````python
# Write your solution here

def column_name(col):
  name = ""
  col += 1

  # will turn columb numbers into letters, makes sense, but is alot. 
  while col > 0:
    col, remainder = divmod(col - 1, 26)
    name = chr(65 + remainder) + name

  return name


def draw_labeled_board(board):
  n = len(board)
  cols = len(board[0])
  width = max(3, len(column_name(cols - 1)))
  row_width = len(str(n))
  space = " " * (row_width + 1)

  header = space + " "
  for col in range(cols):
    header += column_name(col).center(width) + " "
  print(header)

  border = space + ("+" + "-" * width) * cols + "+"

  for row in range(n):
    print(border)
    line = str(row + 1).rjust(row_width) + " |" # needed to look up what this was

    for cell_ in board[row]:
      if cell_ == 0:
        symbol = " "
      elif cell_ == 1:
        symbol = "X"
      else:
        symbol = "O"

      line += symbol.center(width) + "|"

    print(line)

  print(border)
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
# Write your solution here

def place_location(board, player, location):
  location = location.strip().upper()

  split = 0
  while split < len(location) and "A" <= location[split] <= "Z":
    split += 1

  if split == 0:
    return False

  letters = location[:split]

  try:
    row = int(location[split:]) - 1
  except ValueError:
    return False

  col = 0
  for letter in letters:
    col = col * 26 + ord(letter) - ord("A") + 1
  col -= 1

  return place_mark(board, player, row, col)
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
def play_game(n):
  board = create_board(n)
  player = 1

  while check_game(board) == -1:
    draw_labeled_board(board)
    ask_for_move(board, player)
    player = 3 - player

  draw_labeled_board(board)
  result = check_game(board)

  if result == 0:
    print("It is a draw!")
  else:
    print(f"Player {result} won!")

  return board
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Write your solution here

# alrighty, after alot of obersvation of multiple ways of this this via what an LLM said, i think i found the one that makes the most sense
# im sure its not the most effiecient, but it makes sense

import random


def empty_spots(board):
  spots = []

  for row in range(len(board)):
    for col in range(len(board[row])):
      if board[row][col] == 0:
        spots.append((row, col))

  return spots


def simple_computer(board, player):
  spots = empty_spots(board)

  if check_game(board) != -1:
    return None

  for test_player in [player, 3 - player]:
    for row, col in spots:
      board[row][col] = test_player
      result = check_game(board)
      board[row][col] = 0

      if result == test_player:
        return row, col

  return random.choice(spots)


def play_computer(n, computer):
  board = create_board(n)
  player = 1

  while check_game(board) == -1:
    draw_labeled_board(board)

    if player == 1:
      ask_for_move(board, player)
    else:
      row, col = computer(board, player)
      place_mark(board, player, row, col)
      print("Computer chose", column_name(col) + str(row + 1))

    player = 3 - player

  draw_labeled_board(board)
  result = check_game(board)

  if result == 0:
    print("It is a draw!")
  elif result == 1:
    print("You won!")
  else:
    print("The computer won!")

  return board
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
# Write your solution here
# alrighty, ngl, this on the edge of "yes understanding" and "too complex". like looking at it makes mostly sense, but i would just have never thought of some of this stuff, idk how to explain it. 
def score_board(board, computer):
  n = len(board)
  other = 3 - computer
  lines = []

  for row in board:
    lines.append(row)

  for col in range(n):
    lines.append([board[row][col] for row in range(n)])

  lines.append([board[i][i] for i in range(n)])
  lines.append([board[i][n - 1 - i] for i in range(n)])

  score = 0

  for line in lines:
    if other not in line:
      score += line.count(computer) ** 2

    if computer not in line:
      score -= line.count(other) ** 2

  return score


def minimax(board, player, computer, depth):
  result = check_game(board)
  n = len(board)

  win_score = (2 * n + 2) * n * n + 1

  if result == computer:
    return win_score

  if result == 3 - computer:
    return -win_score

  if result == 0:
    return 0

  if depth == 0:
    return score_board(board, computer)

  scores = []

  for row, col in empty_spots(board):
    board[row][col] = player
    score = minimax(board, 3 - player, computer, depth - 1)
    board[row][col] = 0
    scores.append(score)


  if player == computer:
    return max(scores)

  return min(scores)


def search_computer(board, player, max_depth=2):
  if max_depth < 1:
    raise ValueError("max_depth must be at least 1")

  if check_game(board) != -1:
    return None

  best_score = float("-inf")
  best_moves = []

  for row, col in empty_spots(board):
    board[row][col] = player
    score = minimax(board, 3 - player, player, max_depth - 1)
    board[row][col] = 0

    if score > best_score:
      best_score = score
      best_moves = [(row, col)]

    elif score == best_score:
      best_moves.append((row, col))

  return random.choice(best_moves)


def smarter_computer(board, player):
  return search_computer(board, player, max_depth=2)
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
# Write your solution here
from time import perf_counter


def compare_computers(n, games=10, max_depth=2):
  wins = 0
  losses = 0
  draws = 0
  start = perf_counter()

  for game in range(games):
    board = create_board(n)
    player = 1

    smart_player = 1 if game % 2 == 0 else 2

    while check_game(board) == -1:
      if player == smart_player:
        row, col = search_computer(board, player, max_depth)
      else:
        row, col = simple_computer(board, player)

      place_mark(board, player, row, col)
      player = 3 - player

    result = check_game(board)

    if result == smart_player:
      wins += 1
    elif result == 0:
      draws += 1
    else:
      losses += 1

  seconds = perf_counter() - start

  print(f"{n}x{n}: {wins} wins, {losses} losses, {draws} draws")
  print(f"Smarter computer win rate: {wins / games:.0%}")
  print(f"Time for {games} games: {seconds:.2f} seconds")
  print()

  return wins, losses, draws
````

````python
# Test your solution here

random.seed(3402)

for n in [3, 4, 5]:
  compare_computers(n, games=10, max_depth=2)
````
